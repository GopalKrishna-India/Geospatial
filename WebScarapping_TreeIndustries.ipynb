```python
# ============================================================
# GOOGLE MAPS INDUSTRY SCRAPER FOR INDIA
# ------------------------------------------------------------
# Extracts:
# 1. Plywood factories
# 2. Biomass plants
# 3. Paper mills
# 4. Veneer industries
# 5. Agro-processing units
#
# OUTPUT:
# CSV with:
# - industry_type
# - state
# - name
# - address
# - rating
# - user_ratings_total
# - latitude
# - longitude
# - place_id
#
# REQUIREMENTS:
# pip install googlemaps pandas tqdm
#
# IMPORTANT:
# You need Google Places API Key
# Enable:
# - Places API
# - Geocoding API
#
# ============================================================

import googlemaps
import pandas as pd
import time
from tqdm import tqdm

# ============================================================
# 1. GOOGLE API KEY
# ============================================================

API_KEY = "AIzaSyDhwLg0nb0NIgKND-Iq24lXybGNfnbaZ7k"

gmaps = googlemaps.Client(key=API_KEY)

# ============================================================
# 2. STATES OF INDIA
# ============================================================

states = [
    "Andhra Pradesh",
    "Arunachal Pradesh",
    "Assam",
    "Bihar",
    "Chhattisgarh",
    "Goa",
    "Gujarat",
    "Haryana",
    "Himachal Pradesh",
    "Jharkhand",
    "Karnataka",
    "Kerala",
    "Madhya Pradesh",
    "Maharashtra",
    "Manipur",
    "Meghalaya",
    "Mizoram",
    "Nagaland",
    "Odisha",
    "Punjab",
    "Rajasthan",
    "Sikkim",
    "Tamil Nadu",
    "Telangana",
    "Tripura",
    "Uttar Pradesh",
    "Uttarakhand",
    "West Bengal"
]

# ============================================================
# 3. INDUSTRY SEARCH TERMS
# ============================================================

industry_queries = [
    "plywood factory",
    "biomass power plant",
    "paper mill",
    "veneer industry",
    "agro processing unit",
    "saw mill",
    "timber market",
    "wood depot",
    "bamboo industry",
    "fruit processing unit",
    "fodder market",
    "wooden toy industry",
    "furniture cluster"
]

# ============================================================
# 4. STORAGE
# ============================================================

all_results = []

# ============================================================
# 5. SEARCH FUNCTION
# ============================================================

def search_places(query, state):

    full_query = f"{query} in {state}, India"

    print(f"\nSearching: {full_query}")

    try:

        results = gmaps.places(query=full_query)

        while True:

            for place in results.get("results", []):

                item = {}

                item["industry_type"] = query
                item["state"] = state
                item["name"] = place.get("name")
                item["address"] = place.get("formatted_address")
                item["rating"] = place.get("rating")
                item["user_ratings_total"] = place.get("user_ratings_total")
                item["place_id"] = place.get("place_id")

                location = place.get("geometry", {}).get("location", {})

                item["latitude"] = location.get("lat")
                item["longitude"] = location.get("lng")

                all_results.append(item)

            # ------------------------------------------------
            # NEXT PAGE
            # ------------------------------------------------

            next_page_token = results.get("next_page_token")

            if not next_page_token:
                break

            time.sleep(3)

            results = gmaps.places(
                query=full_query,
                page_token=next_page_token
            )

    except Exception as e:

        print(f"ERROR: {e}")

# ============================================================
# 6. RUN SEARCH
# ============================================================

for state in tqdm(states):

    for query in industry_queries:

        search_places(query, state)

        time.sleep(2)

# ============================================================
# 7. DATAFRAME
# ============================================================

df = pd.DataFrame(all_results)

# ============================================================
# 8. REMOVE DUPLICATES
# ============================================================

df = df.drop_duplicates(subset=["place_id"])

# ============================================================
# 9. SAVE CSV
# ============================================================

output_csv = "india_tree_based_industries_googlemaps.csv"

df.to_csv(output_csv, index=False)

print("\n================================================")
print("DONE")
print("Total Records:", len(df))
print("Saved:", output_csv)
print("================================================")
```
